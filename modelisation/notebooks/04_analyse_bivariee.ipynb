# Analyse bivariée

Comparaison descriptive des features au défaut, limitée aux lignes entraînables.

```python
from pathlib import Path
import pandas as pd

jeu = pd.read_parquet(Path.cwd() / 'data/processed/jeu_socle.parquet').query('cible.notna()').copy()
jeu['cible'] = jeu['cible'].astype(int)
jeu.groupby('tendance_epargne_12m', observed=False)['cible'].agg(['count', 'mean'])
```
